Run `python behavior/extract_clips.py` first. Every number below is computed from the real annotation file.

```python
import sys, os
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT)); os.chdir(ROOT)
import config, pandas as pd, numpy as np
import matplotlib.pyplot as plt
```

```python
ann = pd.read_csv(config.PROCESSED_DIR / 'annotations.csv')
seg = pd.read_csv(config.PROCESSED_DIR / 'segments.csv')
print('videos:', ann.video_key.nunique(), ' annotated actions:', len(ann), ' training windows:', len(seg))
ann.head()
```

```python
ann['duration_s'] = (ann.end_frame - ann.start_frame + 1) / ann.fps
ann.groupby('class_name')['duration_s'].describe()
```

```python
seg.pivot_table(index='class_name', columns='split', values='segment_id', aggfunc='count', fill_value=0).plot(kind='bar', figsize=(8,4))
plt.title('Windows per class and split'); plt.tight_layout(); plt.show()
```
