# 02 - Preprocessing
How annotated actions become fixed windows (see `behavior/extract_clips.py`).
- shorter than MIN_SEG_SEC: dropped
- longer than MAX_SEG_SEC: cut in pieces
- videos are split as a whole (no leakage)

```python
import sys, os
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT)); os.chdir(ROOT)
import config, pandas as pd, numpy as np
import matplotlib.pyplot as plt
```

```python
print('MIN_SEG_SEC', config.MIN_SEG_SEC, 'MAX_SEG_SEC', config.MAX_SEG_SEC, 'SEQ_LEN', config.SEQ_LEN, 'TARGET_FPS', config.TARGET_FPS)
seg = pd.read_csv(config.PROCESSED_DIR / 'segments.csv')
seg['sec'] = (seg.end_frame - seg.start_frame + 1) / seg.fps
seg.sec.hist(bins=30); plt.xlabel('window length (s)'); plt.show()
```

```python
# leakage check: each video must be in exactly one split
print((seg.groupby('video_key').split.nunique() == 1).all())
```
